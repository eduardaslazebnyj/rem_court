# Build IUROPA judgment-level node features

This section constructs a **one-row-per-judgment** node table for later relational event modelling (REM) and exports it as `node_features.xlsx`.

Design principles:

- The node universe is IUROPA **judgments** (`decision_type == "judgment"`) across all courts by default.
- Time-varying features (age, CJEU tenure, prior assignments, professional experience, novelty) are evaluated **as of the judgment date** and never use later observations.
- One-to-many IUROPA tables are aggregated before merging so that the result remains one row per `decision_id`.
- Network-history variables such as indegree, PageRank, citation momentum, etc. are deliberately **not** included here.
- The workbook contains the node table plus a feature dictionary and missingness summary.

You can restrict the universe later (for example, `Court of Justice` only) by changing the configuration below.

```python
# ============================================================
# CONFIGURATION + INPUT CHECKS
# ============================================================

import numpy as np
import pandas as pd
import re
from collections import defaultdict
from functools import lru_cache
from pathlib import Path

# Default REM node universe: all judgments in IUROPA.
# Example restriction: COURTS = {"Court of Justice"}
COURTS = None
NODE_DECISION_TYPES = {"judgment"}

# Metadata-based novelty uses only earlier judgments.
# Set False if you want a faster first run.
RUN_NOVELTY = True

OUTPUT_XLSX = Path("node_features.xlsx")

required_core = {
    "decisions",
    "proceedings",
    "procedures",
    "judge_assignments",
    "advocate_general_assignments",
    "judges_advocates_general",
    "appointments",
}

missing_core = sorted(required_core - set(dfs))
if missing_core:
    raise FileNotFoundError(
        "Missing core IUROPA tables: " + ", ".join(missing_core)
        + ". Put the CSVs in the same folder as this notebook and rerun the import cell above."
    )

optional_tables = [
    "decisions_curated", "proceedings_curated", "cases", "hearings", "referrals",
    "joins", "appeals", "transfers", "chamber_assignments", "biographies",
    "judge_advocate_general_careers_curated", "directory_codes", "keywords",
    "headnotes", "parties", "observers", "national_courts"
]

print("Core tables available.")
print("Optional tables available:")
for t in optional_tables:
    print(f"  {t:45s} {'YES' if t in dfs else 'NO'}")
```

```python
# ============================================================
# HELPERS
# ============================================================

def slugify(x):
    """Safe, stable column-name fragment."""
    x = str(x).strip().lower()
    x = re.sub(r"[^a-z0-9]+", "_", x)
    return x.strip("_") or "missing"


def unique_join(series, sep="; "):
    vals = [str(x).strip() for x in series.dropna() if str(x).strip()]
    return sep.join(sorted(set(vals))) if vals else np.nan


def n_unique_nonmissing(series):
    return series.dropna().nunique()


def bool_any(series):
    s = pd.to_numeric(series, errors="coerce").fillna(0)
    return int((s != 0).any())


def first_nonmissing(series):
    s = series.dropna()
    return s.iloc[0] if len(s) else np.nan


def excel_safe_value(x):
    """Excel cells cannot contain strings longer than 32,767 characters."""
    if isinstance(x, str) and len(x) > 32760:
        return x[:32757] + "..."
    return x


def merge_one_to_one(left, right, on="decision_id", how="left"):
    """Merge and assert that the node table remains one row per decision."""
    before = len(left)
    out = left.merge(right, on=on, how=how)
    if len(out) != before:
        raise ValueError(
            f"Merge on {on!r} changed row count from {before:,} to {len(out):,}. "
            "Aggregate the right-hand table first."
        )
    return out


def add_count_pivots(df, index, col, prefix):
    """Return wide count dummies such as party_type__company_count."""
    if col not in df.columns or df.empty:
        return pd.DataFrame({index: df[index].drop_duplicates()})
    tmp = df.dropna(subset=[index, col]).copy()
    if tmp.empty:
        return pd.DataFrame(columns=[index])
    p = pd.crosstab(tmp[index], tmp[col])
    p.columns = [f"{prefix}__{slugify(c)}_count" for c in p.columns]
    return p.reset_index()


def prior_count_by_actor_date(df, actor_col, date_col, out_col):
    """
    Number of earlier rows for the actor, using only dates strictly before the current date.
    Rows on the same date do not count one another.
    """
    tmp = df[[actor_col, date_col]].dropna().copy()
    daily = (
        tmp.groupby([actor_col, date_col], as_index=False)
           .size()
           .rename(columns={"size": "n_on_date"})
           .sort_values([actor_col, date_col])
    )
    daily[out_col] = daily.groupby(actor_col)["n_on_date"].cumsum() - daily["n_on_date"]
    return daily[[actor_col, date_col, out_col]]


def interval_union_days(intervals, asof_date):
    """Union length of date intervals clipped at asof_date (avoids double counting overlaps)."""
    asof = pd.Timestamp(asof_date)
    clean = []
    for start, end in intervals:
        if pd.isna(start):
            continue
        start = pd.Timestamp(start)
        if start > asof:
            continue
        end = asof if pd.isna(end) else min(pd.Timestamp(end), asof)
        if end < start:
            continue
        clean.append((start, end))
    if not clean:
        return np.nan
    clean.sort(key=lambda z: z[0])
    merged = [list(clean[0])]
    for start, end in clean[1:]:
        if start <= merged[-1][1] + pd.Timedelta(days=1):
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    return sum((end - start).days + 1 for start, end in merged)


def descriptive_missingness(df):
    return (
        pd.DataFrame({
            "feature": df.columns,
            "n_missing": [df[c].isna().sum() for c in df.columns],
            "pct_missing": [100 * df[c].isna().mean() for c in df.columns],
            "n_unique": [df[c].nunique(dropna=True) for c in df.columns],
            "dtype": [str(df[c].dtype) for c in df.columns],
        })
        .sort_values(["pct_missing", "feature"], ascending=[False, True])
        .reset_index(drop=True)
    )
```

```python
# ============================================================
# 1. BASE NODE UNIVERSE: ONE ROW PER JUDGMENT
# ============================================================

decisions = dfs["decisions"].copy()
decisions["decision_date"] = pd.to_datetime(decisions["decision_date"], errors="coerce")

nodes = decisions[
    decisions["decision_type"].astype(str).str.lower().isin({x.lower() for x in NODE_DECISION_TYPES})
].copy()

if COURTS is not None:
    nodes = nodes[nodes["court"].isin(COURTS)].copy()

nodes = nodes.drop_duplicates("decision_id").reset_index(drop=True)
nodes["decision_year"] = nodes["decision_date"].dt.year

# Use the curated decision table when available because it already links decisions -> proceedings -> cases.
if "decisions_curated" in dfs:
    dc = dfs["decisions_curated"].copy()
    keep = [c for c in [
        "decision_id", "case_id", "case_number", "proceeding_id", "proceeding_number",
        "proceeding_suffix", "proceeding_name"
    ] if c in dc.columns]
    dc = dc[keep].drop_duplicates("decision_id")
    nodes = merge_one_to_one(nodes, dc, on="decision_id")
else:
    proceedings = dfs["proceedings"].copy()
    p_keep = [c for c in [
        "proceeding_id", "case_id", "proceeding_number", "year", "suffix", "proceeding_name"
    ] if c in proceedings.columns]
    p = proceedings[p_keep].rename(columns={
        "proceeding_id": "lead_proceeding_id",
        "year": "proceeding_year",
        "suffix": "proceeding_suffix"
    })
    nodes = merge_one_to_one(nodes, p, on="lead_proceeding_id")

    if "cases" in dfs and "case_id" in nodes.columns:
        cases = dfs["cases"].copy()
        c_keep = [c for c in ["case_id", "case_number", "year"] if c in cases.columns]
        c = cases[c_keep].rename(columns={"year": "case_year"}).drop_duplicates("case_id")
        nodes = merge_one_to_one(nodes, c, on="case_id")

# Normalize the proceeding key used below.
if "proceeding_id" not in nodes.columns and "lead_proceeding_id" in nodes.columns:
    nodes["proceeding_id"] = nodes["lead_proceeding_id"]

# The year a proceeding was lodged is available in proceedings; it is not an exact filing date.
proc = dfs["proceedings"].copy()
if "year" in proc.columns:
    proc_year = proc[["proceeding_id", "year"]].drop_duplicates("proceeding_id").rename(
        columns={"year": "proceeding_lodged_year"}
    )
    nodes = merge_one_to_one(nodes, proc_year, on="proceeding_id")
    nodes["approx_years_lodging_to_decision"] = (
        nodes["decision_year"] - pd.to_numeric(nodes["proceeding_lodged_year"], errors="coerce")
    )

print(f"Judgment nodes: {len(nodes):,}")
print(nodes[[c for c in ["decision_id", "decision_date", "court", "case_number", "proceeding_name"] if c in nodes]].head())
```

```python
# ============================================================
# 2. PROCEDURES + OUTCOMES
# ============================================================

procedures = dfs["procedures"].copy()

proc_agg = procedures.groupby("decision_id", as_index=False).agg(
    procedures=("procedure", unique_join),
    n_procedures=("procedure", n_unique_nonmissing),
    outcomes=("outcome", unique_join),
    n_outcomes=("outcome", n_unique_nonmissing),
)

nodes = merge_one_to_one(nodes, proc_agg)

# Wide procedure and outcome counts retain the full coding rather than imposing a hand-picked taxonomy.
proc_pivot = add_count_pivots(procedures, "decision_id", "procedure", "procedure")
outcome_pivot = add_count_pivots(procedures, "decision_id", "outcome", "outcome")
nodes = merge_one_to_one(nodes, proc_pivot)
nodes = merge_one_to_one(nodes, outcome_pivot)

# A few convenient high-level flags.
proc_text = nodes["procedures"].fillna("").str.lower()
nodes["is_preliminary_ruling"] = proc_text.str.contains("preliminary").astype(int)
nodes["is_appeal"] = proc_text.str.contains("appeal").astype(int)
nodes["is_annulment"] = proc_text.str.contains("annulment").astype(int)
nodes["is_infringement"] = (
    proc_text.str.contains("failure to fulfill") | proc_text.str.contains("failure to fulfil")
).astype(int)
nodes["is_damages_action"] = proc_text.str.contains("damages").astype(int)
nodes["is_failure_to_act"] = proc_text.str.contains("failure to act").astype(int)
```

```python
# ============================================================
# 3. PROCEEDING STRUCTURE: JOINS, APPEALS, TRANSFERS
# ============================================================

# The curated proceedings table is the cleanest source when available.
if "proceedings_curated" in dfs:
    pc = dfs["proceedings_curated"].copy()
    keep = [c for c in [
        "proceeding_id", "is_main_proceeding", "joined_proceeding_count",
        "joined_proceeding_list", "is_transferred", "transferred_to_proceeding_id",
        "appeal_count", "appeal_list", "is_appeal_of", "appeal_of_list",
        "decision_count", "has_judgment", "has_ag_opinion"
    ] if c in pc.columns]
    pc = pc[keep].drop_duplicates("proceeding_id")
    nodes = merge_one_to_one(nodes, pc, on="proceeding_id")

    if "joined_proceeding_count" in nodes:
        nodes["is_joined_case"] = (pd.to_numeric(nodes["joined_proceeding_count"], errors="coerce").fillna(0) > 0).astype(int)
        nodes["n_proceedings_in_bundle"] = 1 + pd.to_numeric(nodes["joined_proceeding_count"], errors="coerce").fillna(0)
else:
    # Fallback: aggregate joins directly.
    if "joins" in dfs:
        joins = dfs["joins"].copy()
        j = joins.groupby("lead_proceeding_id", as_index=False).agg(
            joined_proceeding_count=("joined_proceeding_id", "nunique"),
            joined_proceeding_list=("joined_proceeding_id", unique_join),
        ).rename(columns={"lead_proceeding_id": "proceeding_id"})
        nodes = merge_one_to_one(nodes, j, on="proceeding_id")
        nodes["joined_proceeding_count"] = nodes["joined_proceeding_count"].fillna(0)
        nodes["is_joined_case"] = (nodes["joined_proceeding_count"] > 0).astype(int)
        nodes["n_proceedings_in_bundle"] = 1 + nodes["joined_proceeding_count"]

# Note: "appeal_count" in proceedings_curated can refer to later appeals and is therefore
# descriptive, but should NOT be used as a predictor in an ex-ante model of the original judgment.
# By contrast, "is_appeal_of" / the appeal procedure itself is already known at the judgment date.
```

```python
# ============================================================
# 4. HEARINGS + CASE-TIMING FEATURES
# ============================================================

if "hearings" in dfs:
    hearings = dfs["hearings"].copy()
    hearings["hearing_date"] = pd.to_datetime(hearings["hearing_date"], errors="coerce")
    h = hearings.groupby("decision_id", as_index=False).agg(
        n_hearings=("hearing_date", "count"),
        first_hearing_date=("hearing_date", "min"),
        last_hearing_date=("hearing_date", "max"),
    )
    nodes = merge_one_to_one(nodes, h)
    nodes["days_first_hearing_to_decision"] = (nodes["decision_date"] - nodes["first_hearing_date"]).dt.days
    nodes["days_last_hearing_to_decision"] = (nodes["decision_date"] - nodes["last_hearing_date"]).dt.days

# Preliminary-reference timing is handled with the referral metadata below because those
# dates are exact when reported, unlike the generic proceeding year.
```

```python
# ============================================================
# 5. CHAMBER / FORMATION
# ============================================================

if "chamber_assignments" in dfs:
    chambers = dfs["chamber_assignments"].copy()
    ch = chambers.groupby("decision_id", as_index=False).agg(
        chambers=("chamber", unique_join),
        n_chambers=("chamber", n_unique_nonmissing),
    )
    nodes = merge_one_to_one(nodes, ch)
    chamber_text = nodes["chambers"].fillna("").str.lower()
    nodes["is_grand_chamber"] = chamber_text.str.contains("grand chamber").astype(int)
    nodes["is_full_court_or_plenary"] = (
        chamber_text.str.contains("plenary") | chamber_text.str.contains("full court")
    ).astype(int)
```

## Judicial personnel features

The following cells construct features **at the judgment date**:

- approximate age from `birth_year`;
- exact cumulative CJEU service time from appointment intervals, clipped at the judgment date and unioned so overlapping appointments do not double count;
- prior judge / rapporteur / AG assignments using only decisions on **strictly earlier dates**;
- panel-level age, tenure, experience, gender and Member-State diversity;
- dated professional-background measures from the biography table.

The biography code is intentionally conservative: where start/end years are missing, it does not invent a duration.

```python
# ============================================================
# 6. PREPARE PERSON MASTER + APPOINTMENT-HISTORY FUNCTIONS
# ============================================================

people = dfs["judges_advocates_general"].copy()
appointments = dfs["appointments"].copy()
appointments["start_date"] = pd.to_datetime(appointments["start_date"], errors="coerce")
appointments["end_date"] = pd.to_datetime(appointments["end_date"], errors="coerce")

# Store appointment records by actor for fast as-of calculations.
appointment_records = {}
for actor_id, g in appointments.groupby("actor_id"):
    appointment_records[actor_id] = g[["court", "position", "start_date", "end_date"]].to_dict("records")

@lru_cache(maxsize=None)
def _service_days_cached(actor_id, asof_date_str, court, position_key):
    asof = pd.Timestamp(asof_date_str)
    positions = None if position_key == "__ANY__" else set(position_key.split("||"))
    intervals = []
    for r in appointment_records.get(actor_id, []):
        if court != "__ANY__" and r["court"] != court:
            continue
        if positions is not None and r["position"] not in positions:
            continue
        intervals.append((r["start_date"], r["end_date"]))
    return interval_union_days(intervals, asof)


def service_years(actor_id, asof_date, court=None, positions=None):
    if pd.isna(actor_id) or pd.isna(asof_date):
        return np.nan
    court_key = "__ANY__" if court is None or pd.isna(court) else str(court)
    pos_key = "__ANY__" if positions is None else "||".join(sorted(positions))
    d = _service_days_cached(str(actor_id), pd.Timestamp(asof_date).strftime("%Y-%m-%d"), court_key, pos_key)
    return np.nan if pd.isna(d) else d / 365.25

# First CJEU appointment date is fixed historical information.
first_appointment = (
    appointments.groupby("actor_id", as_index=False)["start_date"].min()
    .rename(columns={"start_date": "first_cjeu_appointment_date"})
)
```

```python
# ============================================================
# 7. JUDGES, RAPPORTEUR + PANEL FEATURES
# ============================================================

ja = dfs["judge_assignments"].copy()
ja = ja.merge(nodes[["decision_id", "decision_date", "court"]], on="decision_id", how="inner")
ja = ja.merge(people, left_on="judge_id", right_on="actor_id", how="left", suffixes=("", "_person"))
ja = ja.merge(first_appointment, on="actor_id", how="left")

# Age at judgment date (birth year only -> approximate age).
ja["judge_age_approx"] = ja["decision_date"].dt.year - pd.to_numeric(ja["birth_year"], errors="coerce")

# Cumulative service up to the judgment date.
JUDGE_LIKE_POSITIONS = {"judge", "President", "Vice-President"}
ja["judge_cjeu_service_years_any_role"] = ja.apply(
    lambda r: service_years(r["judge_id"], r["decision_date"]), axis=1
)
ja["judge_judicial_service_years"] = ja.apply(
    lambda r: service_years(r["judge_id"], r["decision_date"], positions=JUDGE_LIKE_POSITIONS), axis=1
)
ja["judge_current_court_judicial_service_years"] = ja.apply(
    lambda r: service_years(r["judge_id"], r["decision_date"], court=r["court"], positions=JUDGE_LIKE_POSITIONS), axis=1
)

# Experience measured as the number of earlier decision assignments.
prior_judge = prior_count_by_actor_date(ja, "judge_id", "decision_date", "prior_judge_assignments")
ja = ja.merge(prior_judge, on=["judge_id", "decision_date"], how="left")

# Prior assignments specifically as Judge-Rapporteur.
rapporteur_hist = ja[ja["is_judge_rapporteur"].astype(bool)].copy()
prior_rap = prior_count_by_actor_date(
    rapporteur_hist, "judge_id", "decision_date", "prior_rapporteur_assignments"
)
ja = ja.merge(prior_rap, on=["judge_id", "decision_date"], how="left")
ja["prior_rapporteur_assignments"] = ja["prior_rapporteur_assignments"].fillna(0)

# Panel aggregates.
panel = ja.groupby("decision_id", as_index=False).agg(
    panel_judge_ids=("judge_id", unique_join),
    panel_judge_names=("display_label", unique_join),
    panel_size=("judge_id", "nunique"),
    panel_mean_age_approx=("judge_age_approx", "mean"),
    panel_median_age_approx=("judge_age_approx", "median"),
    panel_min_age_approx=("judge_age_approx", "min"),
    panel_max_age_approx=("judge_age_approx", "max"),
    panel_age_sd=("judge_age_approx", "std"),
    panel_share_female=("is_female", "mean"),
    panel_n_member_states=("member_state", n_unique_nonmissing),
    panel_member_states=("member_state", unique_join),
    panel_mean_cjeu_service_years=("judge_cjeu_service_years_any_role", "mean"),
    panel_mean_judicial_service_years=("judge_judicial_service_years", "mean"),
    panel_mean_current_court_service_years=("judge_current_court_judicial_service_years", "mean"),
    panel_mean_prior_judge_assignments=("prior_judge_assignments", "mean"),
    panel_median_prior_judge_assignments=("prior_judge_assignments", "median"),
)
panel["panel_member_state_diversity"] = panel["panel_n_member_states"] / panel["panel_size"].replace(0, np.nan)
nodes = merge_one_to_one(nodes, panel)

# Rapporteur-specific node attributes (rare multiple rapporteurs are aggregated rather than dropped).
rap = ja[ja["is_judge_rapporteur"].astype(bool)].copy()
rap_agg = rap.groupby("decision_id", as_index=False).agg(
    rapporteur_ids=("judge_id", unique_join),
    rapporteur_names=("display_label", unique_join),
    n_rapporteurs=("judge_id", "nunique"),
    rapporteur_birth_year=("birth_year", "mean"),
    rapporteur_age_approx=("judge_age_approx", "mean"),
    rapporteur_is_female=("is_female", "mean"),
    rapporteur_member_states=("member_state", unique_join),
    rapporteur_cjeu_service_years=("judge_cjeu_service_years_any_role", "mean"),
    rapporteur_judicial_service_years=("judge_judicial_service_years", "mean"),
    rapporteur_current_court_service_years=("judge_current_court_judicial_service_years", "mean"),
    rapporteur_prior_judge_assignments=("prior_judge_assignments", "mean"),
    rapporteur_prior_rapporteur_assignments=("prior_rapporteur_assignments", "mean"),
)
nodes = merge_one_to_one(nodes, rap_agg)

print("Judge/panel features added.")
```

```python
# ============================================================
# 8. ADVOCATE-GENERAL FEATURES
# ============================================================

aga = dfs["advocate_general_assignments"].copy()
aga = aga.merge(nodes[["decision_id", "decision_date", "court"]], on="decision_id", how="inner")
aga = aga.merge(people, left_on="advocate_general_id", right_on="actor_id", how="left", suffixes=("", "_person"))
aga = aga.merge(first_appointment, on="actor_id", how="left")

aga["ag_age_approx"] = aga["decision_date"].dt.year - pd.to_numeric(aga["birth_year"], errors="coerce")
aga["ag_cjeu_service_years_any_role"] = aga.apply(
    lambda r: service_years(r["advocate_general_id"], r["decision_date"]), axis=1
)
aga["ag_role_service_years"] = aga.apply(
    lambda r: service_years(r["advocate_general_id"], r["decision_date"], positions={"Advocate-General"}), axis=1
)
aga["ag_current_court_role_service_years"] = aga.apply(
    lambda r: service_years(
        r["advocate_general_id"], r["decision_date"], court=r["court"], positions={"Advocate-General"}
    ), axis=1
)

prior_ag = prior_count_by_actor_date(aga, "advocate_general_id", "decision_date", "ag_prior_assignments")
aga = aga.merge(prior_ag, on=["advocate_general_id", "decision_date"], how="left")

ag_agg = aga.groupby("decision_id", as_index=False).agg(
    ag_ids=("advocate_general_id", unique_join),
    ag_names=("display_label", unique_join),
    n_advocates_general=("advocate_general_id", "nunique"),
    ag_birth_year=("birth_year", "mean"),
    ag_age_approx=("ag_age_approx", "mean"),
    ag_is_female=("is_female", "mean"),
    ag_member_states=("member_state", unique_join),
    ag_cjeu_service_years=("ag_cjeu_service_years_any_role", "mean"),
    ag_role_service_years=("ag_role_service_years", "mean"),
    ag_current_court_role_service_years=("ag_current_court_role_service_years", "mean"),
    ag_prior_assignments=("ag_prior_assignments", "mean"),
)

nodes = merge_one_to_one(nodes, ag_agg)
nodes["has_advocate_general"] = nodes["n_advocates_general"].fillna(0).gt(0).astype(int)

print("AG features added.")
```

```python
# ============================================================
# 9. DATED PROFESSIONAL BACKGROUND FOR RAPPORTEUR + AG
# ============================================================

# These measures use only biography entries with a known start year no later than the judgment.
# Duration measures use only entries with BOTH start and end years known. Missing dates are not imputed.

BIO_TYPES = ["lawyer", "civil servant", "academic", "politician", "judge", "other"]

if "biographies" in dfs:
    bios = dfs["biographies"].copy()
    bios = bios[bios["entry_type"].astype(str).str.lower().eq("position")].copy()
    bios["start_year"] = pd.to_numeric(bios["start_year"], errors="coerce")
    bios["end_year"] = pd.to_numeric(bios["end_year"], errors="coerce")

    # civil servant; lawyer is intentionally split into both categories.
    bio_rows = []
    for _, r in bios.iterrows():
        ptypes = [x.strip() for x in str(r.get("position_type", "")).split(";") if x.strip() and x != "nan"]
        for pt in ptypes:
            rr = r.copy()
            rr["position_type_split"] = pt
            bio_rows.append(rr)
    bios_long = pd.DataFrame(bio_rows) if bio_rows else pd.DataFrame()

    bio_by_actor = {aid: g.copy() for aid, g in bios_long.groupby("actor_id")} if not bios_long.empty else {}

    @lru_cache(maxsize=None)
    def bio_summary(actor_id, asof_year):
        out = {}
        g = bio_by_actor.get(actor_id)
        for pt in BIO_TYPES:
            out[f"bio_has_{slugify(pt)}_by_date"] = 0
            out[f"bio_n_{slugify(pt)}_positions_started_by_date"] = 0
            out[f"bio_known_years_{slugify(pt)}_by_date"] = 0.0
        if g is None or pd.isna(asof_year):
            return out
        y = int(asof_year)
        for pt in BIO_TYPES:
            x = g[g["position_type_split"].eq(pt)].copy()
            x_started = x[x["start_year"].notna() & (x["start_year"] <= y)]
            out[f"bio_has_{slugify(pt)}_by_date"] = int(len(x_started) > 0)
            out[f"bio_n_{slugify(pt)}_positions_started_by_date"] = int(len(x_started))

            # Conservative known-duration measure: require both endpoints.
            xd = x_started[x_started["end_year"].notna()].copy()
            if not xd.empty:
                intervals = []
                for _, rr in xd.iterrows():
                    a = int(rr["start_year"])
                    b = min(int(rr["end_year"]), y)
                    if b >= a:
                        intervals.append((a, b))
                # union integer years to avoid double counting concurrent positions of the same type
                covered = set()
                for a, b in intervals:
                    covered.update(range(a, b + 1))
                out[f"bio_known_years_{slugify(pt)}_by_date"] = float(len(covered))
        return out

    def attach_bio_features(person_rows, id_col, prefix):
        if person_rows.empty:
            return pd.DataFrame(columns=["decision_id"])
        rows = []
        for _, r in person_rows[["decision_id", id_col, "decision_date"]].drop_duplicates().iterrows():
            d = bio_summary(str(r[id_col]), r["decision_date"].year if pd.notna(r["decision_date"]) else np.nan)
            d = {f"{prefix}_{k}": v for k, v in d.items()}
            d["decision_id"] = r["decision_id"]
            rows.append(d)
        tmp = pd.DataFrame(rows)
        # Multiple AGs / rapporteurs are rare. For counts/years use mean; for flags use max.
        if tmp.empty:
            return pd.DataFrame(columns=["decision_id"])
        agg_map = {}
        for c in tmp.columns:
            if c == "decision_id":
                continue
            agg_map[c] = "max" if "_has_" in c else "mean"
        return tmp.groupby("decision_id", as_index=False).agg(agg_map)

    rap_bio = attach_bio_features(rap, "judge_id", "rapporteur")
    ag_bio = attach_bio_features(aga, "advocate_general_id", "ag")
    nodes = merge_one_to_one(nodes, rap_bio)
    nodes = merge_one_to_one(nodes, ag_bio)

    print("Dated biography features added.")
else:
    print("biographies.csv not found; biography features skipped.")
```

```python
# ============================================================
# 10. SUBJECT MATTER: DIRECTORY CODES, KEYWORDS, HEADNOTES
# ============================================================

if "directory_codes" in dfs:
    d = dfs["directory_codes"].copy()
    directory_agg = d.groupby("decision_id", as_index=False).agg(
        directory_codes=("code", unique_join),
        n_directory_codes=("code", n_unique_nonmissing),
        directory_sections=("section_label", unique_join),
        n_directory_sections=("section_label", n_unique_nonmissing),
        directory_major_labels=("major_label", unique_join),
        n_directory_major_labels=("major_label", n_unique_nonmissing),
        directory_minor_labels=("minor_label", unique_join),
        n_directory_minor_labels=("minor_label", n_unique_nonmissing),
    )
    nodes = merge_one_to_one(nodes, directory_agg)
    section_pivot = add_count_pivots(d, "decision_id", "section_label", "directory_section")
    nodes = merge_one_to_one(nodes, section_pivot)

if "keywords" in dfs:
    kw = dfs["keywords"].copy()
    kw_agg = kw.groupby("decision_id", as_index=False).agg(
        keywords=("keyword", unique_join),
        n_keywords=("keyword", n_unique_nonmissing),
    )
    nodes = merge_one_to_one(nodes, kw_agg)

if "headnotes" in dfs:
    hn = dfs["headnotes"].copy()
    hn_agg = hn.groupby("decision_id", as_index=False).agg(
        headnotes=("headnote", unique_join),
        n_headnotes=("headnote", "count"),
    )
    nodes = merge_one_to_one(nodes, hn_agg)
```

```python
# ============================================================
# 11. METADATA NOVELTY USING ONLY EARLIER JUDGMENTS
# ============================================================

# Novelty here is NOT a network statistic. It compares a judgment's legal metadata with
# judgments that already existed before its decision date.
#
# For a feature-set S_i:
#   new_share      = share of i's values never observed in an earlier judgment
#   max_jaccard    = maximum Jaccard similarity to any earlier judgment
#   novelty        = 1 - max_jaccard
#
# Same-day judgments do NOT count as prior judgments.
# Directory codes have a pre-/post-Lisbon classification break; interpret exact-code novelty
# around 2009 with caution. Section/major-label novelty is provided as a more stable alternative.

def temporal_set_novelty(node_frame, feature_frame, value_col, prefix):
    base = node_frame[["decision_id", "decision_date"]].dropna(subset=["decision_date"]).copy()
    fs = (
        feature_frame.dropna(subset=["decision_id", value_col])
        .groupby("decision_id")[value_col]
        .apply(lambda s: set(str(x) for x in s.dropna()))
        .to_dict()
    )

    previous_sets = {}
    inverted = defaultdict(set)
    seen_values = set()
    results = []

    for date, day in base.sort_values("decision_date").groupby("decision_date", sort=True):
        day_results = []
        for decision_id in day["decision_id"]:
            s = fs.get(decision_id, set())
            if not s:
                day_results.append({
                    "decision_id": decision_id,
                    f"{prefix}_new_share": np.nan,
                    f"{prefix}_max_jaccard_to_prior": np.nan,
                    f"{prefix}_novelty": np.nan,
                })
                continue

            new_share = len(s - seen_values) / len(s)
            candidates = set()
            for val in s:
                candidates.update(inverted.get(val, set()))

            if candidates:
                max_j = 0.0
                for other in candidates:
                    os = previous_sets[other]
                    denom = len(s | os)
                    if denom:
                        max_j = max(max_j, len(s & os) / denom)
            else:
                max_j = 0.0

            day_results.append({
                "decision_id": decision_id,
                f"{prefix}_new_share": new_share,
                f"{prefix}_max_jaccard_to_prior": max_j,
                f"{prefix}_novelty": 1.0 - max_j,
            })

        # Only AFTER scoring the full day do today's judgments enter history.
        for decision_id in day["decision_id"]:
            s = fs.get(decision_id, set())
            if not s:
                continue
            previous_sets[decision_id] = s
            seen_values.update(s)
            for val in s:
                inverted[val].add(decision_id)

        results.extend(day_results)

    return pd.DataFrame(results)

if RUN_NOVELTY:
    if "directory_codes" in dfs:
        d = dfs["directory_codes"]
        for value_col, prefix in [
            ("code", "directory_code"),
            ("section_label", "directory_section"),
            ("major_label", "directory_major"),
        ]:
            if value_col in d.columns:
                nov = temporal_set_novelty(nodes, d, value_col, prefix)
                nodes = merge_one_to_one(nodes, nov)

    if "keywords" in dfs and "keyword" in dfs["keywords"].columns:
        nov = temporal_set_novelty(nodes, dfs["keywords"], "keyword", "keyword")
        nodes = merge_one_to_one(nodes, nov)

    print("Metadata novelty features added.")
else:
    print("RUN_NOVELTY=False; novelty calculation skipped.")
```

```python
# ============================================================
# 12. PARTIES + OBSERVERS / EX-ANTE SALIENCE PROXIES
# ============================================================

if "parties" in dfs:
    p = dfs["parties"].copy()
    party_agg = p.groupby("decision_id", as_index=False).agg(
        party_ids=("party_id", unique_join),
        party_types=("party_type", unique_join),
        party_roles=("party_role", unique_join),
        n_parties=("party_id", n_unique_nonmissing),
        n_party_types=("party_type", n_unique_nonmissing),
    )
    nodes = merge_one_to_one(nodes, party_agg)
    nodes = merge_one_to_one(nodes, add_count_pivots(p, "decision_id", "party_type", "party_type"))
    nodes = merge_one_to_one(nodes, add_count_pivots(p, "decision_id", "party_role", "party_role"))

    # Standardized party IDs make this a reasonably robust convenience flag.
    commission_party = (
        p.assign(_is_commission=p["party_id"].astype(str).str.contains("commission", case=False, na=False))
         .groupby("decision_id", as_index=False)["_is_commission"].max()
         .rename(columns={"_is_commission": "has_commission_party"})
    )
    commission_party["has_commission_party"] = commission_party["has_commission_party"].astype(int)
    nodes = merge_one_to_one(nodes, commission_party)

if "observers" in dfs:
    o = dfs["observers"].copy()
    obs_agg = o.groupby("decision_id", as_index=False).agg(
        observer_ids=("observer_id", unique_join),
        observer_types=("observer_type", unique_join),
        n_observers=("observer_id", n_unique_nonmissing),
        n_observer_types=("observer_type", n_unique_nonmissing),
    )
    nodes = merge_one_to_one(nodes, obs_agg)
    nodes = merge_one_to_one(nodes, add_count_pivots(o, "decision_id", "observer_type", "observer_type"))

    commission_obs = (
        o.assign(_is_commission=o["observer_id"].astype(str).str.contains("commission", case=False, na=False))
         .groupby("decision_id", as_index=False)["_is_commission"].max()
         .rename(columns={"_is_commission": "has_commission_observer"})
    )
    commission_obs["has_commission_observer"] = commission_obs["has_commission_observer"].astype(int)
    nodes = merge_one_to_one(nodes, commission_obs)
```

```python
# ============================================================
# 13. REFERRALS + NATIONAL-COURT CHARACTERISTICS
# ============================================================

if "referrals" in dfs:
    r = dfs["referrals"].copy()
    r["date_referred"] = pd.to_datetime(r["date_referred"], errors="coerce")
    r["date_received"] = pd.to_datetime(r["date_received"], errors="coerce")

    # Enrich with national-court metadata when available.
    if "national_courts" in dfs:
        nc = dfs["national_courts"].copy()
        nc_key = None
        for candidate in ["national_court_id", "iuropa_national_court_id"]:
            if candidate in nc.columns:
                nc_key = candidate
                break
        if nc_key is not None and "national_court_id" in r.columns:
            nc_keep = [c for c in [
                nc_key, "court_location", "in_hierarchy", "court_level", "court_prestige",
                "court_collegiality", "court_competence", "court_competence_detailed"
            ] if c in nc.columns]
            nc2 = nc[nc_keep].drop_duplicates(nc_key).rename(columns={nc_key: "national_court_id"})
            r = r.merge(nc2, on="national_court_id", how="left")

    ref_agg_dict = {
        "national_court_id": [("referring_court_ids", unique_join), ("n_referring_courts", n_unique_nonmissing)],
        "member_state": [("referring_member_states", unique_join), ("n_referring_member_states", n_unique_nonmissing)],
        "date_referred": [("first_date_referred", "min"), ("last_date_referred", "max")],
        "date_received": [("first_date_received", "min"), ("last_date_received", "max")],
    }

    # Named aggregation assembled dynamically to tolerate version differences.
    named = {}
    for col, specs in ref_agg_dict.items():
        if col in r.columns:
            for out_name, fun in specs:
                named[out_name] = (col, fun)
    for col in ["court_level", "court_prestige", "court_collegiality", "court_competence", "court_competence_detailed"]:
        if col in r.columns:
            named[f"referring_{col}s"] = (col, unique_join)
            named[f"n_referring_{col}s"] = (col, n_unique_nonmissing)

    ref_agg = r.groupby("decision_id", as_index=False).agg(**named)
    ref_count = r.groupby("decision_id", as_index=False).size().rename(columns={"size": "n_referrals"})
    ref_agg = ref_agg.merge(ref_count, on="decision_id", how="left")
    nodes = merge_one_to_one(nodes, ref_agg)

    for col in ["court_level", "court_prestige", "court_collegiality", "court_competence"]:
        if col in r.columns:
            nodes = merge_one_to_one(nodes, add_count_pivots(r, "decision_id", col, f"referring_{col}"))

    # Exact timing for preliminary-reference cases where the judgment reports the dates.
    if "first_date_referred" in nodes:
        nodes["days_first_referral_to_decision"] = (nodes["decision_date"] - nodes["first_date_referred"]).dt.days
        nodes["days_last_referral_to_decision"] = (nodes["decision_date"] - nodes["last_date_referred"]).dt.days
    if "first_date_received" in nodes:
        nodes["days_first_receipt_to_decision"] = (nodes["decision_date"] - nodes["first_date_received"]).dt.days
        nodes["days_last_receipt_to_decision"] = (nodes["decision_date"] - nodes["last_date_received"]).dt.days

    if "first_hearing_date" in nodes and "first_date_received" in nodes:
        nodes["days_first_receipt_to_first_hearing"] = (
            nodes["first_hearing_date"] - nodes["first_date_received"]
        ).dt.days
```

## Features deliberately not included yet

The node workbook intentionally leaves out **endogenous citation-network history**, including indegree, PageRank, prior citation momentum, HITS scores, two-paths, etc. Those can be added later as time-varying REM covariates.

It also does not calculate semantic-text novelty because the CSV bundle shown here contains headnotes/keywords but not a full judgment-text table. If you later add the IUROPA text corpus, a clean extension is an embedding-based novelty score comparing each judgment only with earlier judgments.

```python
# ============================================================
# 14. FINAL CLEANUP + SANITY CHECKS
# ============================================================

# Reassert one row per node.
if nodes["decision_id"].duplicated().any():
    dupes = nodes.loc[nodes["decision_id"].duplicated(False), "decision_id"].head(20).tolist()
    raise ValueError(f"Node table is not unique by decision_id. Examples: {dupes}")

# Put key identifiers first; leave all dynamically constructed columns afterwards.
key_cols = [c for c in [
    "decision_id", "decision_date", "decision_year", "decision_type", "court", "ecli", "celex",
    "case_id", "case_number", "proceeding_id", "lead_proceeding_id", "proceeding_number",
    "proceeding_suffix", "proceeding_name"
] if c in nodes.columns]
other_cols = [c for c in nodes.columns if c not in key_cols]
nodes = nodes[key_cols + other_cols].sort_values(["decision_date", "decision_id"]).reset_index(drop=True)

# Convert any accidental set/list objects to strings for Excel.
for c in nodes.columns:
    if nodes[c].dtype == "object":
        nodes[c] = nodes[c].map(
            lambda x: "; ".join(map(str, x)) if isinstance(x, (list, tuple, set)) else x
        )
        nodes[c] = nodes[c].map(excel_safe_value)

missingness = descriptive_missingness(nodes)

print(f"Final node table: {nodes.shape[0]:,} judgments x {nodes.shape[1]:,} features")
print(f"Unique decision IDs: {nodes['decision_id'].nunique():,}")
display(nodes.head())
display(missingness.head(30))
```

```python
# ============================================================
# 15. FEATURE DICTIONARY
# ============================================================

feature_descriptions = {
    "decision_id": "IUROPA unique decision identifier; primary node key.",
    "decision_date": "Date the judgment was published.",
    "decision_year": "Calendar year of the judgment.",
    "court": "Court issuing the judgment.",
    "case_number": "Court case number linked to the decision.",
    "proceeding_name": "Proceeding name from the Registry/curated linkage.",
    "approx_years_lodging_to_decision": "Decision year minus proceeding lodged year; approximate because IUROPA proceedings provides lodging year, not a universal exact filing date.",
    "n_hearings": "Number of recorded hearings for the decision.",
    "days_first_hearing_to_decision": "Days from first recorded hearing to judgment.",
    "days_last_hearing_to_decision": "Days from last recorded hearing to judgment.",
    "is_grand_chamber": "1 if chamber assignment contains Grand Chamber.",
    "panel_size": "Number of distinct judges assigned to the judgment.",
    "panel_mean_age_approx": "Mean approximate judge age at judgment date, using birth year.",
    "panel_mean_cjeu_service_years": "Mean cumulative CJEU service time of panel judges up to judgment date.",
    "panel_mean_prior_judge_assignments": "Mean number of strictly earlier judge assignments among panel members.",
    "rapporteur_age_approx": "Approximate age of Judge-Rapporteur at judgment date, using birth year.",
    "rapporteur_cjeu_service_years": "Cumulative CJEU service of Judge-Rapporteur up to judgment date.",
    "rapporteur_prior_judge_assignments": "Earlier judge assignments of Judge-Rapporteur before judgment date.",
    "rapporteur_prior_rapporteur_assignments": "Earlier Judge-Rapporteur assignments before judgment date.",
    "ag_age_approx": "Approximate age of assigned Advocate-General at judgment date, using birth year.",
    "ag_cjeu_service_years": "Cumulative CJEU service of assigned AG up to judgment date.",
    "ag_prior_assignments": "Earlier AG assignments before judgment date.",
    "directory_code_novelty": "1 minus maximum Jaccard overlap in exact directory codes with any strictly earlier judgment; use cautiously around the Lisbon directory break.",
    "directory_section_novelty": "1 minus maximum Jaccard overlap in directory sections with any strictly earlier judgment.",
    "keyword_novelty": "1 minus maximum Jaccard overlap in EUR-Lex keywords with any strictly earlier judgment.",
    "n_parties": "Number of unique standardized parties in the judgment.",
    "n_observers": "Number of unique written observers; applies mainly to preliminary rulings.",
    "n_referring_courts": "Number of distinct referring national courts linked to the judgment.",
    "days_first_referral_to_decision": "Days from earliest reported national referral date to judgment.",
    "days_first_receipt_to_decision": "Days from earliest reported Court receipt date to judgment.",
}


def infer_feature_group(c):
    if c.startswith("rapporteur_"):
        return "Judge-Rapporteur"
    if c.startswith("ag_") or c == "has_advocate_general":
        return "Advocate-General"
    if c.startswith("panel_"):
        return "Panel"
    if c.startswith("procedure") or c.startswith("outcome") or c.startswith("is_"):
        return "Procedure / case structure"
    if c.startswith("directory") or c.startswith("keyword") or c.startswith("headnote") or c.startswith("n_directory") or c.startswith("n_keyword") or c.startswith("n_headnote"):
        return "Subject matter / novelty"
    if c.startswith("party") or c.startswith("n_part") or c.startswith("has_commission_party"):
        return "Parties"
    if c.startswith("observer") or c.startswith("n_observer") or c.startswith("has_commission_observer"):
        return "Observers"
    if c.startswith("referring_") or "referral" in c or "receipt" in c or c.startswith("n_referr"):
        return "Referrals / national courts"
    if "hearing" in c or "lodging" in c:
        return "Timing"
    if c in key_cols:
        return "Identifiers"
    return "Other"

feature_dictionary = pd.DataFrame({
    "feature": nodes.columns,
    "group": [infer_feature_group(c) for c in nodes.columns],
    "description": [feature_descriptions.get(c, "Constructed directly from the named IUROPA field/table; see notebook cell and column prefix for definition.") for c in nodes.columns],
    "time_safe_for_ex_ante_model": [
        "NO - future information; descriptive only" if c in {"appeal_count", "appeal_list"}
        else "YES / contemporaneous or historical"
        for c in nodes.columns
    ]
})

display(feature_dictionary.head(40))
```
